Sometimes you'll want to store multiple dictionaries in a list , or a list od items as a value in a dictionary , This is called nesting . you can nest dictionarues  inside a list , a list of items inside a dictionary or even a dictionary inside another  dictionary . The alien_0 dictionaries contains a varity of informations about a second alien, much lessa screen full of aliens . Fpr example . the Following code builds a list of three aliens .



```python
alien_0 = {'color':"green",'points':5}
alien_1 = {'color':"yellow",'points':25}
alien_2 = {'color':"red",'points':15}

aliens = [alien_0,alien_1,alien_2]
for alien in aliens :
  print(alien)
```

These aliens all have the same characteristics , but Pythons consider each one a seperate object , which allows us to modify each aliens individually .
```python
for alien in aliens[4:]:
  if alien['color'] == 'green':
    alien['color'] = 'yellow'
    alien['speed'] = 'medium-fast'
    alien['points']= 10

# showing the first 8 aliens
for alien in aliens[:9]:
  print(alien)
print("....")

# again making  the hard option with red color and point is increase with the level 

for alien in aliens[8:]:
  if alien['color'] == 'yellow':
    alien['color'] = 'red'
    alien['speed'] = 'hard'
    alien['points'] +=5
     

  print(alien)

 

  
  

```

# A List in a Dictionary :

Rather than putting a dictionary inside a list , it's  sometimes usefull to put a list inside a dictionary . For Example , consider  howe you might describe a pizza that someone is ordering , if you were to use only a list , all you cound really store is a list of the pizza's toppings . With a dictionary , a list of topping can be just one aspect of the pizza you're describing .

In a Following example , two kinds of information are stored for each pizza , a types of crust and a list of toppings , The list of toppings is a value associated with the key 'toppings'. To use the items in the list , we give the name of the dictionary and the key 'toppings' as we would any value in the dictionary , instead of returing a single value , we get a list of toppings:

```python
pizza = {
  'crust':'thick',
  'topping':["mushrooms","extra cheese"]
}
print(f"You ordered a {pizza['crust']}-crust pizza"
      ' with the following toppings :')
for toppings in pizza['topping']:
  print(f"\t {toppings}")
```

```python
# another example 

fav_language = {
  'jin':['python','java','R'],
  'sanjay':['java script','HTML','CSS'],
  'sandhaya':['flutter','php','python']
}

for names , languages in fav_language.items():
  print(f"\n {names.title()}, favorite langauge are :")

  for language in languages:
    print(f"\t {language.title()}")
 
```

You should not nest lista and dictionaries too deeply , if you're nesting much deeper than what you see in the precedding example , or if you're working with someone else's code with significants level of nestings, there's most likely a simpler way to solve the problem .
