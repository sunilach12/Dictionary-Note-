```python
alien_0 = {"x_position":1,"y_position":25,"Speed":"medium"}
print(f"Original Position : {alien_0["x_position"]}")

# Move the alien to the  right 
# Determine how far to move the alien based on its currnt speed .

if alien_0["Speed"] =="slow":
  x_increment = 1
elif alien_0["Speed"] == "medium":
  x_increment = 2
else:
  x_increment = 3

# The  new position is the old position plus the increment .

alien_0["x_position"] = alien_0["x_position"] + x_increment
print(f"New Position : {alien_0["x_position"]}")
```

We start by defining an alien with an initial x posit8ion and y position , and a speed  of "medium" .an if-elif-else chain determines how far the alien should move to the right , and assigns this values to the variable x_increment . if the alien's speed is slow , it moves one unit to the right ; if the speed is "medium  it moves two unit to the right ; and if it is "fast , it moves three units to the  right . ### Using get() to Access values :

Using key in square brackets to retrive the value you're intrested in from a dictionary might cause one potential problem : is the key you ask for doesnot exist,you'll get error 

```python
Std_details = {
  "Name":"Sanjit",
  "Age": 19,
  "Country":"Nepal",
}
print(Std_details["Collage"])
```

For dictionaries specifically , you can use the get() method to set a default value that will be returned if the requested key doesn't exist.

The `get()` method requires a key as a first argument , As a second optional argumant , you can pass the value to be returned if the key does't exist.

```python
Collage = Std_details.get("Collage","No  collage avilable ")
print(Collage)
```

If the key "Collage" exis in the dictionary , you'll get the corresponding values . if doesnot , you get the default value . In this case , Point  Doenot exist , and we get a clean meaasge instead of an error  see in above example 

It the require data is avilable it give the data . like the following example .

```python
name = Std_details.get("Name","No name avilable")
print(name)
```
