# Looping through a Dictionary : 
A single Pythons dictionary can contains just a few key-value pairs or millions of pairs . because a dictionary can contains large amount of data , Pythons lets you loop through a dictionary . Dictionary can be used be used to store  information in a variety of ways . Therefore , several different ways exist to loop through them . You can loop through  them . You can loop through all of dictionary' key-value  pairs, through its  keys , or through its values .

### Looping through  ALL key-value pairs .

Before we explore   the different approaches to looping . Let's consder a new dictionary designed  to store information  about a user  on a website . The following dictionary would store one person's username , firt name , and last name .

```python
user_0 = {
  "username": "alexsharma12",
  "first_name":"Alex",
  "last_name":"Sharma"
}
print(user_0)
```

 using the Loop for accessing the key and value

### Another example of the Favourite language:


```python
favorite_language = {
  "sanjay":"JAVA",
  "Nitish":"Python",
  "Brishab":"Java Scripts"
}

for name , language in favorite_language.items():
  print(f"{name.title()}'s favorite language is {language.title()}")
```

print(f"{name.title()}'s favorite language is {language.title()}.")

name, and the value is assigned to the variable language. These descriptive 
names make it much easier to see what the print() call is doing.

# Looping Through a Dictionary's Keys in a Particular order .

Looping Through a dictionary returns the items in the same order they were inserted . Sometimes , Through , you'll want to loop through a dictionarty in a different order. One way to do this is to sort the keys as they're returned in the fpr loop . You can use the sorted() function to get a copy af the keys in order :
 

```python
for name in sorted(favorite_language.keys()):
  print(f"{name.title()} , Thank you for taking the poll.")
```

# Looping Through all Values in a dictionary .

if you are primarily interested in the values that a dictionary contains, you can you the `values()` method to return a sequence of values without any keys . For Example , Say we simply want  a list of all language chosen in our programming language poll, without the name of the person who chose each language .


```python
fov_language = {
  "susan":"JAVA",
  "sanjay":"C++",
  "sharad":"Python",
  "Brishab":"Python"
}

print(f"The following languages have been mentioned :")
for value in fov_language.values():
  print(value.title())

```

The for statement here pulls each value from the dictionary and assigns 
it to the variable language. When these values are printed, we get a list of all 
chosen languages:

This approach pulls all the values from the dictionary without checking for repeats . This might works fine with a small number of values , but in a poll with a large number of respondents , it would result in  a repetitive list ,. To see a=each language chooden without repeatation , we can use a set . A set is a collection in which each item  must be unique.

```python
for values in set(fov_language.values()):
  print(values)
```
